```python
from subprocess import run
import argparse
```

```python
parser = argparse.ArgumentParser()
parser.add_argument('problem_ids', nargs='+', type=int, help='problem IDs to run completions for')
args = parser.parse_args()

problem_ids = args.problem_ids
print(f"will generate completions for problem IDs: {problem_ids}")
```

```python
for problem_id in problem_ids:
    for prompt_detail in ['L', 'M', 'H']:
        print(f"running prompt detail '{prompt_detail}':")
        for t in [0.1, 0.3, 0.5, 0.8, 1.0]:
            print(f"running completions for t={t}")
            for i in range(100):
                with open(f"problem{problem_id}/completions_gpt/completion_{prompt_detail}_t{t}_{i}.v", "w") as f:
                    p = run([f'cat problem{problem_id}/prompt_{prompt_detail}.txt | sgpt --role vgen --temperature {t} "Finish this implementation"'], stdout=f, shell=True)
                    assert p.returncode == 0
```
